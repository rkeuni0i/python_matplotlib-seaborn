```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("./data/KOBIS_역대_박스오피스.csv", encoding="utf-8")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
rate = (df['배급사'].isnull().sum() / df['매출액'].shape[0])*100
rate
```

```python
df2= df.copy()
df.isnull().sum()
```

```python
df2.head()
```

```python
df2.drop('대표국적',axis=1, inplace=True)
df2
```

```python
df2.dropna(subset=['배급사'], inplace=True)
df2
```

```python
df2.isnull().sum()
df2['매출액'].fillna(0, inplace=True)
df2.isnull().sum()
```

```python
product = df2.groupby(['배급사']).sum()
product.sort_values(by='스크린수', ascending=False).head()
```

```python
product[product['스크린수'] >=30000]
```

```python
product = product.sort_values(by='관객수', ascending=False)
product.head(1)
```

```python
df2['스크린수_cut'] = pd.cut(df2['스크린수'],bins=10)
df2['스크린수_cut'].value_counts()
```

```python
df2['스크린수_qcut'] = pd.qcut(df2['스크린수'], q = 10)
df2['스크린수_qcut'].value_counts()
```

```python
plt.rcParams['font.family'] = 'malgun gothic'
plt.figure(figsize=(14,10))
product.head().스크린수[:].plot(kind="bar", rot=0)
plt.xlabel("영화배급사", fontsize=18)
plt.ylabel("스크린수", fontsize=18)
plt.title("흥행 영화 순위 Top5", color = "green")
plt.show()
```

```python
plt.figure(figsize=(14,10))
plt.plot(product.head().스크린수[:], color="blue")
plt.xlabel("영화배급사",fontsize=18)
plt.ylabel("스크린수",fontsize=18)
plt.title("흥행 영화 순위 Top5", color = "green")
plt.show()
```

```python
df = pd.read_csv("./data/it_product.csv", encoding="utf-8")
df.head()
```

```python
group = df.groupby('product').sum()
group = group.sort_values(by="price", ascending=True)
group.drop(['year'], axis=1, inplace=True)

plt.figure(figsize=(8,5))
sns.barplot(x = group.index, y = group.price, data = group, color = "skyblue", hatch="/", edgecolor = "g")
sns.lineplot(x = group.index, y = group.price, data = group, color="red",linestyle = "--", marker="o", markeredgecolor = 'b')
plt.xlabel("product", fontsize=20)
plt.ylabel("가격", fontsize=18)
plt.title("제품별 총액", fontsize=20, color="g")
plt.show()
```
