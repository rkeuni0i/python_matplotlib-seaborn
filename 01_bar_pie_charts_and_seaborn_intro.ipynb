```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'malgun gothic'
df = pd.read_csv("./data/supply.csv")
df
```

```python
label = ['고급형','일반형','저가형']
index = np.arange(len(label))
colors = ['r', 'g', 'b']
plt.bar(index, df['가격'], width=0.5, color = colors, edgecolor = 'purple', alpha=0.3)
plt.title("판매량 집계", fontsize=18, color='r')
plt.xlabel("구분", fontsize=10, color = 'b')
plt.ylabel("가격", fontsize=10)
plt.xticks(index, label, fontsize=12, rotation = 90)
plt.grid()
plt.show()
```

```python
y = np.arange(len(label))
plt.barh(y,df['가격'], color = "g", alpha = 0.3, edgecolor = 'r')
plt.yticks(y, label)
#plt.set_xlabel("금액")
#plt.set_ylabel("index")
#plt.set_title("그래프", loc="left")
plt.title("판매량")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'
df = sns.load_dataset("tips")
df.head(5)
```

```python
sns.set_theme(style="whitegrid")
sns.barplot(x="day", y = "total_bill", data=df)
plt.show()
```

```python
sns.set_theme(style="dark")
sns.barplot(x="sex", y = "total_bill", data = df, errwidth=0)
plt.show()
```

```python
sns.set_theme(style="dark")
sns.barplot(x="total_bill", y = "day", data= df, orient="h", errwidth=0)
plt.show()
```

```python
sns.set_theme(style="darkgrid")
sns.barplot(x="total_bill", y= "day", data = df, orient="h", palette="coolwarm", errwidth=0)
plt.show()
```

```python
sns.set_theme(style="white")
sns.catplot(x="sex", y = "total_bill", data = df, kind="bar", errwidth =0, palette="coolwarm")
plt.show()
```

```python
sns.catplot(x="sex", y = "total_bill", data = df, kind="bar", hue="smoker", errwidth=0)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'
df = sns.load_dataset("titanic")
df.head()
```

```python
sns.set_style("whitegrid")
sns.countplot(x="class", hue="who", data = df)
plt.title("타이타닉 등급 승객", fontsize=18)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.violinplot(x=df['who'], y=df['fare'])
plt.title("sex fare", color = 'r')
plt.show()
```

```python
df_size= df.pivot_table(index="class", columns="sex", aggfunc="size")
df_size
```

```python
sns.heatmap(df_size, annot=True, cmap="RdYlGn", fmt="d", annot_kws={'size':15}, vmin=100, vmax=500, linewidth=5)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'

iris_df = sns.load_dataset("iris")
iris_df.head()
```

```python
sns.jointplot(x="sepal_length", y="sepal_width", data = iris_df, kind="scatter")
plt.suptitle("꽃받침 length and width", y = 1.05, fontsize=18, color = "r")
plt.show()
```

```python
%matplotlib inline
fg = sns.FacetGrid(data=iris_df, col="species")
fg.map(plt.hist, 'sepal_length')
plt.show()
```

```python
sns.pairplot(iris_df, hue="species")
plt.title("Types of iris", fontsize=18)
plt.show()
```

```python
pg = sns.PairGrid(iris_df[["sepal_length", "sepal_width", "petal_length", "petal_width"]])
pg.map(sns.scatterplot, color = 'y', alpha = 0.3)
plt.title("Types of iris", fontsize=18)
plt.show()
```

```python
pg= sns.PairGrid(iris_df[['sepal_length', 'sepal_width','petal_length','petal_width']])
pg.map_upper(sns.kdeplot, color="green")
pg.map_lower(sns.scatterplot, color = "r", alpha = 0.3)
pg.map_diag(sns.histplot, color = "k", alpha = 0.4)
plt.title("Types of iris", fontsize=18)
plt.show()
```

```python
<seaborn>의 그래프 종류!
1. barplot
2. barplot인데 orient="h"로 수평 막대 그래프
3. catplot - 요소별 데이터 비교 hue="분리할 요소" #카테고리 별 분류
4. countplot -  카테고리 별 데이터 항목 개수
5. violinplot- 데이터의 최솟값과 최댓값
6. heatmap - 2차원 테이블(pivot테이블)활용해서 정보 색상 표시 - 상관관계 대한 분포도 표시, 값이 클수록 진하게 
7. jointplot -scatter와 bar 합쳐서 그래프 나타낼 때 kind 옵션으로 데이터 모양 바꾸기 가능
8. FaceGrid - 행, 열 방향으로 서로 다른 조건 적용하여 여러개 그래프(서브플롯 그래프)/ 서브플롯에 적용할 그래프 종류는 map()함수로 
9. pairplot - 3차원 이상의 데이터 그래프로 표혀할 때, 숫자형 데이터에 대해서만 가능, 컬럼 간의 관계를 그래프로 
              map_upper(), map_lower(), map_diag()함수로 대각선 기준 위, 아래, 중심의 그래프 지정 가능 
```
