```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'malgun gothic'
df = pd.read_csv("./data/supply.csv")
df
```

```python
label = ['고급형','일반형','저가형']
index = np.arange(len(label))
plt.bar(index, df['가격'])
plt.title("가격그래프")
plt.xlabel("구분")
plt.ylabel("가격")
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.bar(index, df['가격'], width=0.5)
plt.title("판매량 집계")
plt.xticks(index, label)
plt.show()
```

```python
plt.bar(label, df['가격'], width=0.5 ,color = 'green', alpha = 0.3)
plt.title("판매량 집계")
plt.show()
```

```python
color = ['b', 'gray','skyblue']
plt.bar(index, df['가격'], width = 0.5, color = color, edgecolor = 'red', linewidth=0.5)
plt.title("판매량 집계")
plt.show()
```

```python
plt.barh(label, df['가격'])
plt.show()
```

```python
label = ['고급형','일반형','저가형']
plt.pie(df['가격'], labels = label, autopct= "%.2f%%")
plt.title("판매량 집계",fontsize=18)
plt,show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'
df = sns.load_dataset("tips")
df.head()
```

```python
sns.set_theme(style= "dark")
sns.barplot(y="day",x="total_bill", data = df, errwidth = 0, orient="h", palette="coolwarm")
plt.show()
```

```python
sns.catplot(x="sex", y="total_bill", data = df, kind="bar",errwidth=0, hue="smoker")
plt.show()
```

```python
df = sns.load_dataset("titanic")
df.head()
```

```python
sns.set_theme(style="whitegrid")
sns.countplot(x="class", data = df, hue="who")
plt.title("Titanic Class Passenger")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.set_theme(style="whitegrid")
sns.violinplot(x= "who", y="fare", data = df, palette=["b",'orange','green'])
plt.title("Title Fare Description", fontsize=18)
plt.show()
```

```python
df1 = df.pivot_table(index="class", columns = "sex", aggfunc = "size")
sns.heatmap(df1, annot=True,annot_kws={'size':10}, cmap="RdYlGn", fmt="d", linewidth = 10)
plt.title("Heatmap Graph")
plt.show()
```

```python
iris = sns.load_dataset("iris")
iris.head()
```

```python
sns.jointplot(x="sepal_length", y= "sepal_width",data = iris, kind="kde")
plt.show()
```

```python
fg = sns.FacetGrid(data= iris, col = "species")
fg.map(plt.hist, 'sepal_length')
```

```python
pg =sns.PairGrid(iris[['sepal_length','sepal_width','petal_length','petal_width']])
pg.map_upper(sns.kdeplot)
pg.map_lower(sns.scatterplot)
pg.map_diag(sns.histplot)
plt.show()
```
